## Exercise 5: the four bugs
1. `Print` → `print` (case-sensitive)
2. missing closing `"` after `Chia AI`
3. missing comma between `1` and `"of"`
4. the last string starts with `"` but ends with `'`; quotes must match. (The apostrophe in *Let's* also means we should use double quotes.)

```python
print("Welcome to Chia AI")
print("Session", 1, "of", 8)
print("Let's learn Python!")
```
