## 1. Hello, World!

The tradition when learning any programming language is to make the computer say *Hello, World!*

```python
print("Hello, World!")
```

`print` is a **function**: a ready-made action. Whatever you put inside the brackets is printed. Text must be inside quotes: that's called a **string**.

## 2. Single quotes, double quotes

```python
print("Karibu!")
print('Karibu!')
print("It's a sunny day in Mombasa")   # double quotes let us use an apostrophe inside
```

## 3. Printing several items at once

Separate items with commas. `print` adds a space between them. Numbers don't need quotes.

```python
print("My name is", "Achieng")
print("I am", 21, "years old")
print("Year of study:", 2, "| Course:", "BSc. Statistics")
```

## 4. Python as a calculator

```python
print(250 + 180)     # addition
print(1000 - 345)    # subtraction
print(65 * 3)        # multiplication
print(1000 / 4)      # division always gives a decimal
print(2 ** 10)       # power: 2 to the power of 10
```

```python
# A shopping list: 2 packets of unga at KSh 210 and 1 loaf of bread at KSh 65
print("Total to pay: KSh", 2 * 210 + 65)
```

## 5. Comments

Anything after `#` is a note for humans. Python ignores it.

```python
# This line is a comment and does nothing
print("Comments help other people (and future you) understand your code.")  # so does this one
```

## 6. New lines, tabs and multi-line text

```python
print("Line one\nLine two")     # \n = new line
print("Name:\tAmina")           # \t = tab
print()                         # an empty line
print("""
*************************
*   WELCOME TO PYTHON   *
*************************
""")
```
